```python


# ── Safety ───────────────────────────────────────────────────
import os
os.environ["OMP_NUM_THREADS"] = "1"

# ── Imports ──────────────────────────────────────────────────
import random
import numpy as np
import pandas as pd
import librosa
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    f1_score, classification_report, confusion_matrix
)

torch.backends.cudnn.benchmark = True


# ── Reproducibility ──────────────────────────────────────────
def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything()


# ── Device ───────────────────────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using: {device}")


# ── Paths ────────────────────────────────────────────────────
BASE      = "/kaggle/input/competitions/digitrecognition-ee708"
train_csv = f"{BASE}/train.csv"
train_dir = f"{BASE}/train_audio/train_audio"
test_dir  = f"{BASE}/test_audio/test_audio"

train_df = pd.read_csv(train_csv)
test_files = os.listdir(test_dir)
test_ids   = [f.replace(".wav", "") for f in test_files if f.endswith(".wav")]
test_df    = pd.DataFrame({"id": test_ids})

print(f"Train: {len(train_df)} | Test: {len(test_df)}")


# ── Config ───────────────────────────────────────────────────
SR        = 16000   
SAMPLES   = SR      
N_MELS    = 128     
N_MFCC    = 40      

BATCH_SIZE = 64
EPOCHS     = 20
LR         = 1e-3
NUM_CLASSES = 10

cache_dir = "/kaggle/working/mel_cache_v2"
os.makedirs(cache_dir, exist_ok=True)




def add_noise(y, min_snr=15, max_snr=40):
    """Add Gaussian noise at a random SNR between min_snr and max_snr dB."""
    snr_db  = random.uniform(min_snr, max_snr)
    rms_sig = np.sqrt(np.mean(y ** 2)) + 1e-9
    rms_noise = rms_sig / (10 ** (snr_db / 20))
    noise   = np.random.randn(len(y)) * rms_noise
    return (y + noise).astype(np.float32)


def time_stretch(y, min_rate=0.85, max_rate=1.15):
    """Speed up or slow down without changing pitch."""
    rate = random.uniform(min_rate, max_rate)
    y_stretched = librosa.effects.time_stretch(y, rate=rate)
    
    if len(y_stretched) < SAMPLES:
        y_stretched = np.pad(y_stretched, (0, SAMPLES - len(y_stretched)))
    else:
        y_stretched = y_stretched[:SAMPLES]
    return y_stretched


def pitch_shift(y, min_steps=-3, max_steps=3):
    """Shift pitch by a random number of semitones."""
    steps = random.uniform(min_steps, max_steps)
    return librosa.effects.pitch_shift(y, sr=SR, n_steps=steps)


def random_gain(y, min_db=-6, max_db=6):
    """Multiply amplitude by a random gain in dB."""
    gain = 10 ** (random.uniform(min_db, max_db) / 20)
    return np.clip(y * gain, -1.0, 1.0)


def time_cutout(y, max_fraction=0.2):
    """Zero out a random contiguous segment (up to max_fraction of length)."""
    y = y.copy()
    n = int(len(y) * random.uniform(0.05, max_fraction))
    start = random.randint(0, len(y) - n)
    y[start:start + n] = 0.0
    return y


def spec_augment(mel, n_freq_masks=2, n_time_masks=2, freq_width=15, time_width=20):
    """
    SpecAugment (Park et al. 2019):
      - Mask random frequency bands  → forces model to not rely on
        narrow spectral peaks alone
      - Mask random time windows     → forces model to use all parts
        of the spectrogram for classification
    Proportional masking is more robust than a fixed 8-band mask.
    """
    mel = mel.copy()
    n_freq, n_time = mel.shape

    # Frequency masking
    for _ in range(n_freq_masks):
        f  = random.randint(0, freq_width)
        f0 = random.randint(0, n_freq - f)
        mel[f0:f0 + f, :] = 0

    # Time masking
    for _ in range(n_time_masks):
        t  = random.randint(0, time_width)
        t0 = random.randint(0, n_time - t)
        mel[:, t0:t0 + t] = 0

    return mel


def augment_waveform(y):
    """Apply a random subset of time-domain augmentations."""
    if random.random() < 0.5:
        y = add_noise(y)
    if random.random() < 0.4:
        y = time_stretch(y)
    if random.random() < 0.4:
        y = pitch_shift(y)
    if random.random() < 0.4:
        y = random_gain(y)
    if random.random() < 0.3:
        y = time_cutout(y)
    return y




def load_audio(path):
    y, _ = librosa.load(path, sr=SR)
    if len(y) < SAMPLES:
        y = np.pad(y, (0, SAMPLES - len(y)))
    else:
        y = y[:SAMPLES]
    return y.astype(np.float32)


def extract_features(y):
    """
    Returns a (3, N_MELS, T) array:
      channel 0 = log-mel spectrogram
      channel 1 = delta (Δ)
      channel 2 = delta-delta (ΔΔ)
    """
    mel   = librosa.feature.melspectrogram(y=y, sr=SR, n_mels=N_MELS,
                                            n_fft=1024, hop_length=160)
    mel_db = librosa.power_to_db(mel).astype(np.float32)

    delta  = librosa.feature.delta(mel_db, order=1)
    delta2 = librosa.feature.delta(mel_db, order=2)

    return np.stack([mel_db, delta, delta2], axis=0)   # (3,128,T)


# ── Cache builder (avoids re-extracting every run) ───────────
def build_cache(df, audio_dir, name):
    cache_file = os.path.join(cache_dir, f"{name}.npy")

    if os.path.exists(cache_file):
        print(f"Loading cache: {name}")
        return np.load(cache_file, mmap_mode='r')

    print(f"Building cache: {name}")
    data = []
    for _, row in tqdm(df.iterrows(), total=len(df)):
        path = os.path.join(audio_dir, row['id'] + ".wav")
        if not os.path.exists(path):
            continue
        y   = load_audio(path)
        feat = extract_features(y)   # (3, 128, T)
        data.append(feat)

    data = np.stack(data)            # (N, 3, 128, T)
    np.save(cache_file, data)
    return data


# ── Dataset ──────────────────────────────────────────────────
class AudioDataset(Dataset):
    """
    Returns normalised 3-channel features.
    During training, applies waveform-level augmentation
    by re-loading from disk + SpecAugment on the mel channel.
    For speed, we only do SpecAugment on cached features.
    """
    def __init__(self, features, labels=None, ids=None, audio_dir=None, train=True):
        self.features  = features    # (N, 3, 128, T)  — cached, static
        self.labels    = labels
        self.ids       = ids
        self.audio_dir = audio_dir
        self.train     = train

    def __len__(self):
        return len(self.features)

    def __getitem__(self, idx):
        feat = self.features[idx].copy()   # (3, 128, T)

        if self.train:
            # SpecAugment on the mel channel (channel 0)
            feat[0] = spec_augment(feat[0])

        # Per-channel standardisation
        for c in range(feat.shape[0]):
            mu  = feat[c].mean()
            std = feat[c].std() + 1e-6
            feat[c] = (feat[c] - mu) / std

        x = torch.tensor(feat).float()   # (3, 128, T)

        if self.labels is not None:
            return x, torch.tensor(self.labels[idx]).long()
        return x




class SEBlock(nn.Module):
    """
    Squeeze-and-Excitation block.
    Globally averages each channel → FC bottleneck → sigmoid gate
    → rescale original channels.
    Adds <0.5% parameter overhead but consistently lifts accuracy.
    """
    def __init__(self, channels, reduction=8):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        # x: (B, C, H, W)
        s = x.mean(dim=[2, 3])           # (B, C) — squeeze
        s = self.fc(s).unsqueeze(-1).unsqueeze(-1)  # (B, C, 1, 1) — excitation
        return x * s                     # (B, C, H, W) — scale


class ResBlock(nn.Module):
    """
    2-layer residual block with SE attention.
    Input and output have the same number of channels, enabling
    the identity skip connection without a projection layer.
    """
    def __init__(self, channels, dropout=0.1):
        super().__init__()
        self.conv1 = nn.Conv2d(channels, channels, 3, padding=1, bias=False)
        self.bn1   = nn.BatchNorm2d(channels)
        self.conv2 = nn.Conv2d(channels, channels, 3, padding=1, bias=False)
        self.bn2   = nn.BatchNorm2d(channels)
        self.se    = SEBlock(channels)
        self.drop  = nn.Dropout2d(dropout)

    def forward(self, x):
        residual = x
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out = self.se(out)
        out = self.drop(out)
        return F.relu(out + residual)   # skip connection


class DownsampleBlock(nn.Module):
    """
    Stride-2 conv to halve spatial dimensions and double channels.
    Includes a matching 1×1 projection on the residual path.
    """
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv    = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
        )
        self.shortcut = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 1, stride=2, bias=False),
            nn.BatchNorm2d(out_ch),
        )
        self.se = SEBlock(out_ch)

    def forward(self, x):
        return F.relu(self.se(self.conv(x)) + self.shortcut(x))


class AudioCNNGRU(nn.Module):
    """
    Architecture summary:
      Input  : (B, 3, 128, T)   — 3-channel spectrogram
      CNN stem → 4 residual stages with progressive downsampling
      → Adaptive pool to (C, 4, T') to preserve temporal axis
      → Reshape to sequence (B, T', C*4) for GRU
      → Bidirectional GRU → mean pool over time
      → FC head → 10 logits
    """
    def __init__(self, num_classes=10, gru_hidden=256):
        super().__init__()

        # ── Stem: 3 → 32 channels ────────────────────────────
        self.stem = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(),
        )

        # ── Residual stages with downsampling ─────────────────
        self.stage1 = nn.Sequential(DownsampleBlock(32, 64),   ResBlock(64))
        self.stage2 = nn.Sequential(DownsampleBlock(64, 128),  ResBlock(128))
        self.stage3 = nn.Sequential(DownsampleBlock(128, 256), ResBlock(256))

        self.freq_pool = nn.AdaptiveAvgPool2d((4, None))

        gru_in = 256 * 4
        self.gru = nn.GRU(
            input_size=gru_in,
            hidden_size=gru_hidden,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=0.3
        )

        self.head = nn.Sequential(
            nn.Linear(gru_hidden * 2, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        # x: (B, 3, 128, T)
        x = self.stem(x)           # (B, 32, 128, T)
        x = self.stage1(x)         # (B, 64,  64, T/2)
        x = self.stage2(x)         # (B, 128, 32, T/4)
        x = self.stage3(x)         # (B, 256, 16, T/8)

        x = self.freq_pool(x)      # (B, 256, 4, T/8)
        B, C, F, T = x.shape
        x = x.permute(0, 3, 1, 2)          # (B, T, C, F)
        x = x.reshape(B, T, C * F)         # (B, T, 1024)

        x, _ = self.gru(x)                 # (B, T, 512)
        x = x.mean(dim=1)                  # (B, 512) — temporal mean pooling

        return self.head(x)                # (B, 10)


def mixup_batch(x, y, alpha=0.4):
    """
    Mixup regularisation: interpolates pairs of training samples.
    Applied from epoch 1 (not halfway through like the baseline).

    Why it helps: smooths the decision boundary between digits
    that sound similar (e.g. "one" vs "nine", "four" vs "five"),
    preventing over-confident predictions on ambiguous inputs.
    """
    lam   = np.random.beta(alpha, alpha)
    index = torch.randperm(x.size(0)).to(x.device)
    x_mix = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]
    return x_mix, y_a, y_b, lam


class LabelSmoothingLoss(nn.Module):
    """
    Cross-entropy with label smoothing (ε = smoothing).
    Instead of training against hard 0/1 targets, each correct
    class gets (1 - ε) and the remaining ε is spread across
    all classes. This prevents the model from becoming over-confident
    and improves calibration.
    """
    def __init__(self, num_classes, smoothing=0.1):
        super().__init__()
        self.num_classes = num_classes
        self.smoothing   = smoothing

    def forward(self, pred, target):
        log_prob = F.log_softmax(pred, dim=-1)
        nll      = -log_prob.gather(dim=-1, index=target.unsqueeze(1)).squeeze(1)
        smooth   = -log_prob.mean(dim=-1)
        loss     = (1 - self.smoothing) * nll + self.smoothing * smooth
        return loss.mean()


def evaluate(model, loader):
    """Returns macro F1 and full per-class classification report."""
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            preds = torch.argmax(model(x), dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y.cpu().numpy())

    f1 = f1_score(all_labels, all_preds, average='macro')
    report = classification_report(all_labels, all_preds,
                                   target_names=[str(i) for i in range(NUM_CLASSES)])
    return f1, report, all_labels, all_preds


def plot_confusion_matrix(labels, preds, save_path="/kaggle/working/confusion_matrix.png"):
    """
    Confusion matrix shows you *which digits the model confuses*.
    E.g. if row=9 col=5 is hot, the model often predicts 5 when
    the true label is 9. This guides targeted improvements:
    collect more of that digit, add specific augmentations, etc.
    """
    cm = confusion_matrix(labels, preds)
    cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Raw counts
    sns.heatmap(cm, annot=True, fmt='d', ax=axes[0],
                xticklabels=range(10), yticklabels=range(10))
    axes[0].set_title("Confusion matrix (counts)")
    axes[0].set_xlabel("Predicted")
    axes[0].set_ylabel("True")

    # Normalised (recall per class on diagonal)
    sns.heatmap(cm_norm, annot=True, fmt='.2f', ax=axes[1],
                xticklabels=range(10), yticklabels=range(10))
    axes[1].set_title("Confusion matrix (row-normalised = recall)")
    axes[1].set_xlabel("Predicted")
    axes[1].set_ylabel("True")

    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.close()
    print(f"Confusion matrix saved to {save_path}")


train_feats = build_cache(train_df, train_dir, "train_v2")
test_feats  = build_cache(test_df,  test_dir,  "test_v2")

train_dataset = AudioDataset(train_feats, train_df['label'].values, train=True)
eval_dataset  = AudioDataset(train_feats, train_df['label'].values, train=False)
test_dataset  = AudioDataset(test_feats,  None,                     train=False)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE,
                          shuffle=True,  num_workers=4, pin_memory=True)
eval_loader  = DataLoader(eval_dataset,  batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=4)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=4)

model     = AudioCNNGRU(num_classes=NUM_CLASSES).to(device)
criterion = LabelSmoothingLoss(NUM_CLASSES, smoothing=0.1)
optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)

scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)

best_f1       = 0.0
patience      = 5
patience_ctr  = 0
best_path     = "/kaggle/working/best_model.pt"

print(f"\nModel parameters: {sum(p.numel() for p in model.parameters()):,}")
print("=" * 60)

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for x, y in tqdm(train_loader, desc=f"Epoch {epoch+1:02d}/{EPOCHS}"):
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()

        x_mix, y_a, y_b, lam = mixup_batch(x, y, alpha=0.4)
        out  = model(x_mix)
        loss = lam * criterion(out, y_a) + (1 - lam) * criterion(out, y_b)

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()
        total_loss += loss.item()

    scheduler.step()
    avg_loss = total_loss / len(train_loader)

    # ── Evaluation ───────────────────────────────────────────
    f1, report, labels, preds = evaluate(model, eval_loader)
    lr_now = scheduler.get_last_lr()[0]

    print(f"\nEpoch {epoch+1:02d} | loss: {avg_loss:.4f} | F1: {f1:.4f} | LR: {lr_now:.2e}")

    # ── Early stopping + checkpointing ───────────────────────
    if f1 > best_f1:
        best_f1  = f1
        patience_ctr = 0
        torch.save(model.state_dict(), best_path)
        print(f"  ✓ New best model saved (F1={best_f1:.4f})")
    else:
        patience_ctr += 1
        print(f"  No improvement ({patience_ctr}/{patience})")
        if patience_ctr >= patience:
            print("Early stopping triggered.")
            break

print(f"\nBest F1: {best_f1:.4f}")


model.load_state_dict(torch.load(best_path, map_location=device))
f1_final, report_final, labels_final, preds_final = evaluate(model, eval_loader)

print("\n" + "=" * 60)
print(f"FINAL EVAL F1 (macro): {f1_final:.4f}")
print("=" * 60)
print("\nPer-class report:")
print(report_final)

plot_confusion_matrix(labels_final, preds_final)

model.eval()
all_preds = []

with torch.no_grad():
    for x in test_loader:
        x = x.to(device)
        out = model(x)
        all_preds.extend(torch.argmax(out, dim=1).cpu().numpy())

submission = pd.DataFrame({"id": test_df["id"], "label": all_preds})
submission.to_csv("/kaggle/working/submission.csv", index=False)
print("\nsubmission.csv saved — done!")


pred_counts = pd.Series(all_preds).value_counts().sort_index()
print("\nPredicted class distribution:")
print(pred_counts.to_string())
```
